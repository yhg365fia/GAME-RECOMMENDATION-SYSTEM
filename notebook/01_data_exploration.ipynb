```python
meta = meta.dropna(subset=["Name"])

meta_cols = ["Genres", "Tags", "Categories", "About the game"]

meta = meta.dropna(subset=meta_cols, thresh=2)

meta.head()
```

추천 점수 계산 단계에서 메타데이터 개수에 따라 가중치를 적용한다. 메타데이터가 모두 없는 게임은 제거하고, 나머지는 메타데이터 완전성에 따라 추천 점수를 보정한다.

메타데이터 4개 → Weight = 1.0
메타데이터 3개 → Weight = 0.95
메타데이터 2개 → Weight = 0.80
메타데이터 1개 → Weight = 0.60
메타데이터 0개 → 제거

최종 추천 점수 = Cosine Similarity × Weight

```python
meta["Genres"] = meta["Genres"].fillna("")
meta["Categories"] = meta["Categories"].fillna("")
meta["About the game"] = meta["About the game"].fillna("")
meta["Tags"] = meta["Tags"].fillna("")
```

```python
# 메타데이터가 2개 이상 있는 행만 유지
meta = meta.dropna(subset=meta_cols, thresh=3)

meta.head()
meta.info()
```

```python
meta["Name"].duplicated().sum()
```

```python
# AppID 중복 제거
meta.drop_duplicates(subset=["AppID"], inplace=True)
meta[meta["Name"].duplicated(keep=False)].sort_values("Name")
meta["Name"].duplicated().sum()
```

```python
meta.to_csv("../data/meta_preprocessed.csv", index=False)
```

```python
# 사용할 컬럼 합치기
meta["combined_features"] = (
    meta["Genres"].str.replace(",", " ", regex=False) + " " +
    meta["Tags"].str.replace(",", " ", regex=False) + " " +
    meta["Categories"].str.replace(",", " ", regex=False)
)

print(meta["combined_features"].head())

meta.to_csv("../data/meta_preprocessed.csv", index=False)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    lowercase=True
)

tfidf_matrix = tfidf.fit_transform(meta["combined_features"])

print(tfidf_matrix.shape)
```

```python
len(tfidf.get_feature_names_out())
tfidf.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```
